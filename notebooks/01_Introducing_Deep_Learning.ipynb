## 6. Pemanasan Python

Konsep Python yang paling sering dipakai di buku ini: **list**, **loop**, **fungsi**, dan **list comprehension**. Semuanya muncul lagi di Bab 3 saat membangun neural network.

````python
# List: deret angka, misal statistik pertandingan
toes = [8.5, 9.5, 9.9, 9.0]
print('Elemen pertama :', toes[0])
print('Panjang list   :', len(toes))
print('Dua terakhir   :', toes[-2:])

# Loop
total = 0
for t in toes:
    total += t
print('Rata-rata      :', total / len(toes))

# Fungsi
def kali_semua(angka, faktor):
    return [a * faktor for a in angka]      # list comprehension

print('Dikali 0.1     :', kali_semua(toes, 0.1))

# zip: berjalan pada dua list sekaligus
a = [1, 2, 3]
b = [4, 5, 6]
print('Pasangan       :', list(zip(a, b)))
````

````python
# Cikal bakal "weighted sum" di Bab 3: jumlah dari perkalian elemen dua list
def w_sum(a, b):
    assert len(a) == len(b)
    output = 0
    for i in range(len(a)):
        output += a[i] * b[i]
    return output

print('w_sum([1,2,3], [4,5,6]) =', w_sum([1, 2, 3], [4, 5, 6]), '= 1*4 + 2*5 + 3*6')
````

## 7. Pemanasan NumPy & Kekuatan Vektorisasi

**NumPy** menyediakan `np.array` — array numerik yang operasinya dijalankan di level C. Operasi pada seluruh array sekaligus (tanpa loop Python) disebut **vektorisasi**.

````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print('a + b     =', a + b)          # elementwise
print('a * b     =', a * b)          # elementwise
print('a * 10    =', a * 10)         # skalar
print('a.dot(b)  =', a.dot(b))       # dot product = w_sum
print('a.sum()   =', a.sum(), '| a.mean() =', a.mean())

M_ = np.array([[1, 2, 3],
               [4, 5, 6]])
print('\nMatriks shape:', M_.shape)
print('Baris ke-0    :', M_[0])
print('Kolom ke-1    :', M_[:, 1])
print('Transpose shape:', M_.T.shape)
````

````python
# Seberapa cepat vektorisasi? Bandingkan loop Python vs NumPy untuk berbagai ukuran
sizes = [1_000, 10_000, 100_000, 1_000_000]
t_loop, t_np = [], []

for n in sizes:
    x = np.random.rand(n)
    y = np.random.rand(n)
    xl, yl = x.tolist(), y.tolist()

    t0 = time.perf_counter(); w_sum(xl, yl); t_loop.append(time.perf_counter() - t0)
    t0 = time.perf_counter(); x.dot(y);      t_np.append(time.perf_counter() - t0)

for n, tl, tn in zip(sizes, t_loop, t_np):
    print(f'n={n:>9,}  loop={tl*1000:8.2f} ms  numpy={tn*1000:6.3f} ms  (~{tl / tn:5.0f}x)')

plt.figure(figsize=(6, 3.5))
plt.loglog(sizes, t_loop, 'o-', label='loop Python')
plt.loglog(sizes, t_np, 'o-', label='NumPy .dot()')
plt.xlabel('Panjang vektor'); plt.ylabel('Waktu (detik)'); plt.legend()
plt.title('Vektorisasi jauh lebih cepat')
plt.show()
````

Hasilnya **sama**, tetapi NumPy puluhan kali lebih cepat (dan selisihnya makin besar untuk operasi matriks). Inilah alasan semua neural network ditulis dengan operasi **vektor/matriks**.

## 8. Matematika SMA yang Dibutuhkan

Hanya tiga hal:
1. **Perkalian dan penjumlahan** — inti dari prediksi neural network.
2. **Grafik x–y** — untuk memvisualisasikan hubungan input dan output, serta error.
3. **Kemiringan (slope) garis** — intuisi untuk *derivative* (Bab 4): *jika x digeser sedikit, berapa banyak y berubah?*

````python
x = np.linspace(-2, 2, 100)

fig, axes = plt.subplots(1, 3, figsize=(13, 3.2))
axes[0].plot(x, 0.5 * x, label='y = 0.5x')
axes[0].plot(x, 2 * x, label='y = 2x')
axes[0].set_title('Perkalian: bobot = kemiringan garis'); axes[0].legend()

axes[1].plot(x, x ** 2)
axes[1].set_title('Grafik x-y: parabola (bentuk kurva error)')

# Slope parabola di x = 1 adalah 2 (garis singgung)
axes[2].plot(x, x ** 2, label='y = x^2')
axes[2].plot(x, 1 + 2 * (x - 1), '--', label='garis singgung di x=1 (slope 2)')
axes[2].scatter([1], [1], color='C3', zorder=3)
axes[2].set_ylim(-1, 4); axes[2].legend(fontsize=8)
axes[2].set_title('Slope: seberapa curam kurva')
for ax in axes:
    ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
plt.tight_layout(); plt.show()
````

## 9. Sekilas: Neural Network yang Belajar

Sebagai cuplikan Bab 3–4, inilah neural network paling sederhana: **satu input, satu bobot (weight), satu output**. Ia memprediksi peluang sebuah tim menang dari rata-rata jumlah jari kaki (*toes*) pemainnya — contoh khas buku ini.

$$\text{prediction} = \text{input} \times \text{weight}$$

````python
weight = 0.1

def neural_network(input, weight):
    prediction = input * weight
    return prediction

number_of_toes = [8.5, 9.5, 10, 9]
for t in number_of_toes:
    print(f'toes = {t:4}  ->  prediksi peluang menang = {neural_network(t, weight):.2f}')
````

Bobot `0.1` di atas hanya **ditebak**. Neural network yang sesungguhnya **mempelajari** bobot dari data. Berikut cuplikan **gradient descent** (dijelaskan tuntas di Bab 4): network diberi satu contoh (input 8.5, tim **menang** → target 1.0) dan memperbaiki bobotnya sendiri berulang-ulang.

````python
weight, input, goal = 0.5, 8.5, 1.0
alpha = 0.01
history = []

for iteration in range(20):
    pred = input * weight                 # 1. predict
    error = (pred - goal) ** 2            # 2. compare
    weight -= alpha * (pred - goal) * input   # 3. learn
    history.append((iteration, pred, error, weight))

for it, p, e, w in history[:5] + history[-2:]:
    print(f'iter {it:2d}: prediksi={p:.4f}  error={e:.6f}  bobot baru={w:.4f}')

hist = np.array(history)
fig, axes = plt.subplots(1, 2, figsize=(10, 3))
axes[0].plot(hist[:, 0], hist[:, 1], 'o-'); axes[0].axhline(goal, color='C3', ls='--', label='target')
axes[0].set_title('Prediksi mendekati target'); axes[0].set_xlabel('Iterasi'); axes[0].legend()
axes[1].plot(hist[:, 0], hist[:, 2], 'o-'); axes[1].set_yscale('log')
axes[1].set_title('Error turun'); axes[1].set_xlabel('Iterasi')
plt.tight_layout(); plt.show()
````

Dalam beberapa iterasi, prediksi mendekati 1.0 dan error turun drastis. Inilah *fun payoff* yang dijanjikan Trask: **melihat sesuatu yang kita bangun sendiri belajar**.

## Latihan

1. Tulis fungsi `ele_mul(a, b)` (perkalian elemen dua list) dengan loop Python, lalu bandingkan hasilnya dengan `np.array(a) * np.array(b)`.
2. Pada cuplikan neural network yang belajar di atas, ubah targetnya menjadi `0.0` (tim kalah). Ke mana bobot bergerak?

### Contoh solusi

````python
# Latihan 1
def ele_mul(a, b):
    assert len(a) == len(b)
    return [a[i] * b[i] for i in range(len(a))]

a, b = [1, 2, 3], [4, 5, 6]
print('Loop :', ele_mul(a, b))
print('NumPy:', np.array(a) * np.array(b))

# Latihan 2
weight, input, goal = 0.5, 8.5, 0.0
for _ in range(20):
    pred = input * weight
    weight -= 0.01 * (pred - goal) * input
print(f'\nTarget 0 -> bobot akhir {weight:.6f}, prediksi {input * weight:.6f} (bobot turun mendekati 0)')
````
